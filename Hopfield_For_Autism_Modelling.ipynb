```python
import numpy as np

#directions are up down left and right
DIRECTIONS=[(0,1),(1,0),(1,1),(1,-1),]
def in_bounds(x, y, shape):
    return 0 <= x < shape[0] and 0 <= y < shape[1]
def trace_full_line(img, x, y, dx, dy, visited):#extract 12 edges
    seg=[]#start of line
    cx,cy = x, y
    while in_bounds(cx-dx, cy-dy, img.shape) and img[cx-dx, cy-dy] == 1:
        cx-=dx
        cy -=dy
    while in_bounds(cx, cy, img.shape) and img[cx, cy] == 1:#trace forward
        if (cx, cy, dx, dy) in visited:
            break
        seg.append((cx, cy))
        visited.add((cx, cy, dx, dy))
        cx+=dx
        cy+=dy

    return seg


def extract_edges(img, min_len=4):
    visited=set()
    edges=[]

    for x in range(img.shape[0]):
        for y in range(img.shape[1]):
            if img[x,y]!=1:
                continue
            for dx, dy in DIRECTIONS:
                if (x,y,dx,dy) in visited:
                    continue
                seg=trace_full_line(img, x, y, dx, dy, visited)
                if len(seg)>=min_len:
                    edges.append(seg)
    return edges

def is_line_pixel(img,x,y,dx,dy):
    return (in_bounds(x+dx, y+dy,img.shape) and img[x+dx,y+dy]==1 and in_bounds(x-dx, y-dy,img.shape) and img[x-dx,y-dy]==1)


def find_intersections(img):
    intersections=set()
    for x in range(img.shape[0]):
        for y in range(img.shape[1]):
            if img[x,y] != 1:
                continue
            directions_present =[]
            for dx, dy in DIRECTIONS:
                if is_line_pixel(img, x,y, dx, dy):
                    directions_present.append((dx,dy))


            if len(directions_present) >= 2:#it must belong to at least TWO directions to ensure real crossing
                intersections.add((x,y))

    return intersections
def split_edges(edges, intersections):#split edges
    segments = []

    for edge in edges:
        current = [edge[0]]

        for p in edge[1:]:
            current.append(p)

            if p in intersections:
                if len(current) > 1:
                    segments.append(current[:-1])
                current = [p]
        if len(current) > 1:
            segments.append(current)
    return segments



def segment_vector(reference_segments, test_img):

    vec=[]#build vector
    for seg in reference_segments:
        exists = all(test_img[x,y] == 1 for (x,y) in seg)
        vec.append(1 if exists else 0)
    return np.array(vec)


#-------------------------------------------------------------------------------------------------------------------------龍----------------------------------------------


edges=extract_edges(ambiguous)
print("Edges:",len(edges))   #should be 12 for the cube

intersections = find_intersections(ambiguous)

segments = split_edges(edges, intersections)
#remove small junk segments
segments=[s for s in segments if len(s)>=3]

print("segments = ",len(segments))  #is 16 since only 4 lines get cut thus 12+4=16


# vectors
vecfap=segment_vector(segments,fap)
vecfbp=segment_vector(segments, fbp)
vecambiguous= segment_vector(segments,ambiguous)

print("FAP = ", vecfap)
print("FBP = ", vecfbp)
print("Ambiguous is the all 1s since it's the reference ", vecambiguous)
#calculate Hamming distance between the two percepts
hammingdist = sum(a !=b for a,b in zip(vecfap, vecfbp))
print(f"Hamming Distance b/w FAP and FBP =  {hammingdist}")
print(f"Hamming Dist between Ambiguous and FAP is {sum(a!=b for a, b in zip(vecambiguous,vecfap))}")
print(f"Hamming Distance b/w Amb & FBP: {sum(a!=b for a,b in zip(vecambiguous,vecfbp))}")
```

#Find Dwell time, Percept bias and switching rate (reversals)
We also get and plot the PSD which does shows resonance occurs much like the paper Riani et al., 1993

```python
import numpy as np
import math
import random
import matplotlib.pyplot as plt
from scipy.signal import welch
#16 neurons to encode the hpfield figure
N = 16
#convert our patterns to arrays
ambiguous = np.array(vecambiguous)
FAP = np.array(vecfap)
FBP = np.array(vecfbp)

#Calculate the difference to get the direction of oscilatory sine wave (in the double well figure imagine the x axis as this difference and the y as the energy)
diff_pattern = FAP - FBP
#Note:  diff[i] =  1 -> Unique to FAP
# diff[i] = -1 -> Unique to FBP
# diff[i] =  0 -> Shared by both
gamma = 0.15   # ← controls well asymmetry : We can make the wells equal at gamma = 0
bias_vector = gamma * diff_pattern
# Tuning parameters (Adjusted for Hebbian weight scaling)
H0 = 0.5              #amplitude of oscillating input signal for stochastic resonance
bias_strength = 0.1   #How hard the ambiguous input drives the network? (Looking at the input vectors this pushes both equally)
f_signal=0.0023  #Can be varied but I took this value from "Stochastic Resonance in the Perceptual Interpretation of Ambiguous Figures: A Neural Network Model by Riami et al 1994" which is my comparision

W = np.zeros((N, N))   #WEIGHT MATRIX for Hebbian Learning

# Hopfield networks learn best with bipolar states (-1, 1)
#We map our (0, 1) type patterns to (-1, 1) to calculate the synapses.
bipolar_FAP=2*FAP-1
bipolar_FBP =  2*FBP-1

for i in range(N):
    for j in range(N):
        if i==j:
            continue # No self-connections
        #Hebbian rule: W_ij = (1/N) * Sum(Pattern^mu_i * Pattern^mu_j)
        W[i, j] = (bipolar_FAP[i]*bipolar_FAP[j]+bipolar_FBP[i]*bipolar_FBP[j])/N #Outer product

#SIMULATION using weight MATRIX
def run_simulation(T,f_signal,sweeps=200000):
    period=int(round(1.0/f_signal))
    omega =2.0 *math.pi / period
    #the State Vector tracks all 16 neurons
    S=np.copy(FAP)#start perfectly in FAP's basin

    current_state = 1
    samples = np.zeros(sweeps, dtype=np.int8)

    for sweep in range(sweeps):
        H_t = H0 * math.sin(omega * sweep)

        #Asynchronous update per sweep
        for _ in range(N):
            #pick 1 neuron randomly (0 to 15 in index)
            i = random.randrange(N)

            #The Local Field Calculation, basically will it switch?
            # h_i = Network influence + External Ambiguous Bias (the regular necker's cube as an input) + Oscillating Field
            #We can map S_j to bipolar (2*S - 1) for the symmetric network dynamics
            network_influence = np.dot(W[i],(2 * S - 1))
            external_bias=bias_strength*ambiguous[i]

            #we use H_t to only pushes on segments unique to FAP or FBP this pushing it b/w these 2 only
            oscillating_attention = H_t * diff_pattern[i]
            h_i = network_influence + external_bias + oscillating_attention + bias_vector[i]#bias vector is the one with gamm*diff

            # 3. Calculate Energy Change (dE) to flip this specific neuron
            # If S[i] is 1, turning it to 0 means a change (dS) of -1.
            # If S[i] is 0, turning it to 1 means a change (dS) of +1.
            dS = 1 - 2 * S[i]
            dE = -h_i * dS    #standard Hopfield Hamiltonian function

            #Metropolis-Hastings (Glauber Dynamics) flip decision -> https://en.wikipedia.org/wiki/Glauber_dynamics
            if T==0:
                if dE<=0:
                    S[i]=1-S[i]
            else:
                if dE<=0 or random.random()<math.exp(-dE/T):
                    S[i] = 1-S[i]

        #This is reminiscent of physics with a Macroscopic State depending on past patterns (Hysteresis)
        #We now simply count the active states based on the number of "unique" segments of FAP or FBP
        unique_fap_active=np.sum(S[diff_pattern==1])   # Max of 5
        unique_fbp_active = np.sum(S[diff_pattern==-1])  #5 are diff

        #Schmitt Trigger logic - forces b/w 2 percepts
        if unique_fap_active>=4 and unique_fbp_active<=1:
            current_state = 1
        elif unique_fbp_active>=4 and unique_fap_active <= 1:
            current_state=-1

        samples[sweep]=current_state

    burn_in=max(5000, sweeps // 10)
    return samples[burn_in:], period


# POWER SPECTRUM - Used chatgpt for this to use the fourier transform properly
def compute_psd(x, period, f_signal):
    x = x - np.mean(x)
    nperseg = period * 128 # Adjusted to prevent array overflow on shorter sweeps

    freqs, power = welch(x, fs=1.0, nperseg=nperseg, scaling='density')

    peak_bins=set()
    max_harmonics=30

    for h in range(1, max_harmonics + 1):
        target_f =h*f_signal
        bin_idx=np.argmin(np.abs(freqs-target_f))
        peak_bins.update([bin_idx-1,bin_idx,bin_idx+1])

    background = np.copy(power)
    for h in range(1,max_harmonics+1):
        target_f=h*f_signal
        bin_idx=np.argmin(np.abs(freqs - target_f))
        left=max(0, bin_idx-2)
        right=min(len(power)-1, bin_idx + 2)
        background[left:right+1]=np.linspace(background[left],background[right],right-left+1)

    w=11
    smoothed_background=np.convolve(background,np.ones(w)/w,mode='same')

    final_power=np.copy(smoothed_background)#for ease of viewing
    for idx in sorted(list(peak_bins)):
        if 0 <= idx<len(final_power):
            final_power[idx]=power[idx]

    return freqs,final_power

#plot the results of all of this! To show resonance also occurs at certain noise levels

def switching_rate(samples):
    switches = np.sum(np.abs(np.diff(samples))) / 2
    return switches / len(samples)
def dwell_times(samples):
    dwells=[]
    current=samples[0]
    count=1
    for s in samples[1:]:
        if s==current:
            count+=1
        else:
            dwells.append(count)
            count=1
            current=s
    dwells.append(count)
    return np.array(dwells)
def mean_dwell(samples):
    return np.mean(dwell_times(samples))
def percept_bias(samples):
    p_plus = np.mean(samples == 1)
    p_minus = np.mean(samples == -1)
    return [p_plus, p_minus]
def dwell_by_state(samples):
    dwells=dwell_times(samples)
    states=[]
    current=samples[0]
    state_list =[]
    for s in samples[1:]:
        if s!=current:
            state_list.append(current)
            current=s
    state_list.append(current)

    dwells=np.array(dwells)
    state_list=np.array(state_list)

    dw_plus=dwells[state_list == 1]
    dw_minus=dwells[state_list == -1]

    return dw_plus,dw_minus
#temperatures lowered because the Hebbian weights W are smaller in absolute value compared to the Ria97 paper reference before with a manual 0.02 / 0.04 matrix
temperatures = [0.00,0.15, 0.35, 0.60]
spectra = {}

for T in temperatures:
    print(f"running for T = {T}   ...")
    x, period = run_simulation(T,f_signal, sweeps=500000)
    freqs, power = compute_psd(x,period, f_signal)
    spectra[T] = (freqs, power)
    print(f"Switching rate = {switching_rate(x)}")
    print(f"Mean dwell time = {mean_dwell(x)}")
    print(f"Percept bias =  {percept_bias(x)}\n")
    dw_plus, dw_minus = dwell_by_state(x)
    print("Mean dwell (+1 state):",np.mean(dw_plus))
    print("Mean dwell (-1 state):",np.mean(dw_minus))
    dw = dwell_times(x)
    plt.hist(dw, bins=50, density=True)
    plt.xlabel("Dwell time")
    plt.ylabel("Probability")
    plt.title("Dwell time distribution")
    plt.show()

plt.figure(figsize=(8, 5))

for T, (f, p) in spectra.items():
    mask=f>0
    if abs(T - 0.15) < 0.01:
        style ='--'
    elif abs(T - 0.35) < 0.01:
        style=":"
    elif abs(T - 0.60) < 0.01:
        style='-.'
    else:
        style='-'
    plt.loglog(f[mask],p[mask],style, label=f"T = {T}", linewidth=1.5)

plt.xlabel("frequency")
plt.ylabel("P.S.D. (power spectrum)")
plt.title("Power Spectra with the necker's cube vectorised")
plt.legend()
plt.grid(True,which="both", ls="--", alpha=0.5)
plt.xlim(1e-4,5e0)
plt.ylim(1e-5,1e4)
plt.show()
```

```python
#We modify the sim a bit to get it done in a reasonable amount of time
def run_simulation(T, f_signal, sweeps=90000):
    period = int(round(1.0 / f_signal))
    omega = 2.0 * math.pi / period
    S=np.copy(FAP)
    S_bipolar = 2 * S - 1#OPTIM 1 Track the bipolar state to avoid array creation overhead

    #OPTIMIZATION 2:- Pre-compute the exact indices for unique segments to avoid creating boolean mask arrays every single sweep
    idx_fap =np.where(diff_pattern==1)[0]
    idx_fbp =np.where(diff_pattern==-1)[0]
    current_state = 1
    samples = np.zeros(sweeps, dtype=np.int8)

    for sweep in range(sweeps):
        H_t = H0 * math.sin(omega * sweep)
        for _ in range(N):
            i = random.randrange(N)
            network_influence = np.dot(W[i], S_bipolar)
            external_bias = bias_strength * ambiguous[i]#(Now using pre-tracked S_bipolar)
            oscillating_attention = H_t * diff_pattern[i]

            h_i = network_influence + external_bias + oscillating_attention + bias_vector[i]

            #calculate the Energy change which is (dE)
            dS = 1 - 2 * S[i]
            dE = -h_i * dS

            #Metropolis-Hastings flip decision
            if dE <= 0 or random.random() < math.exp(-dE / T):
                S[i] = 1 - S[i]
                # Keep bipolar state perfectly synced with S
                S_bipolar[i] = -S_bipolar[i]

        #Macroscopic State (Now using pre-computed indices)
        unique_fap_active = np.sum(S[idx_fap])
        unique_fbp_active = np.sum(S[idx_fbp])

        if unique_fap_active >= 4 and unique_fbp_active <= 1:
            current_state = 1
        elif unique_fbp_active >= 4 and unique_fap_active <= 1:
            current_state = -1

        samples[sweep] = current_state

    burn_in=max(5000,sweeps//10)
    return samples[burn_in:], period
#psd
def compute_psd(x, period, f_signal):
    x = x - np.mean(x)
    nperseg = period * 128

    # Prevent nperseg from exceeding data length
    if nperseg > len(x):
        nperseg = len(x)
    freqs, power = welch(x, fs=1.0, nperseg=nperseg, scaling='density')
    return freqs, power


#SNR CALC

def compute_snr(freqs, power, f_signal):
    idx = np.argmin(np.abs(freqs - f_signal))

    #noise floor from nearby bins
    window = 15  #go out 15 bins
    bleed = 3    #skip the closest three bins since our signal "bleeds" into them

    noise_indices = list(range(max(0, idx-window), max(0, idx-bleed))) + \
                    list(range(min(len(power), idx+bleed+1), min(len(power), idx+window+1)))

    if len(noise_indices) == 0:
        return 0.0

    noise_floor = np.mean(power[noise_indices])

    #true signal power is the height of the peak - the underlying floor noise
    S = power[idx] - noise_floor

    #convert to the standard decibels(dB)
    if S > 0 and noise_floor>0:
        snr_db=10*np.log10(S / noise_floor)
    else:
        snr_db=0.0

    return snr_db



#the frequencies are from the "Stochastic Resonance in the Perceptual Interpretation of Ambiguous Figures: A Neural Network Model by Riami et al 1994"'s Fig 4
frequencies=[0.0009, 0.0023, 0.01]

#temperatures are lowered to match the new Hebbian energy landscape
temperatures=np.linspace(0.05, 0.7, 25)

results={}  #to store our SNR curves for each of the frequencies

for f_signal in frequencies:
    print(f"\nrunning frequency is = to {f_signal}")
    snr_values = []

    for T in temperatures:
        #running more sweeps for the lower frequencies ensures a cleaner SNR peak
        sweeps_needed = max(100000, int(150 / f_signal))

        x,period = run_simulation(T, f_signal, sweeps=sweeps_needed)

        #If the system never switches (at very low T) we can handle perfectly flat arrays
        if np.all(x == x[0]):
            snr_values.append(0.0)
            continue

        freqs, power = compute_psd(x, period, f_signal)
        snr = compute_snr(freqs, power, f_signal)
        snr_values.append(snr)

    results[f_signal] = snr_values

#plotting
plt.figure(figsize=(8, 6))

markers = ['D','+','s'] # Diamonds,pluses,squares like how they drew their one "Stochastic Resonance in the Perceptual Interpretation of Ambiguous Figures: A Neural Network Model by Riami et al 1994"
for (f_signal, snr_values), marker in zip(results.items(), markers):
    plt.plot(temperatures, snr_values, marker=marker, linestyle='-', linewidth=1.5, label=f"f = {f_signal}")

plt.xlabel("Temperature (T)")
plt.ylabel("Signal-to-Noise Ratio (dB)")
plt.title("Stochastic Resonance in Learned Network (Fig. 4 Equivalent)")
plt.legend()
plt.grid(True, ls="--", alpha=0.5)
plt.show()
```

```python
import numpy as np
import math
import random
import matplotlib.pyplot as plt
from scipy.signal import welch
#16 neurons to encode the hpfield figure
N = 16
#convert our patterns to arrays
ambiguous = np.array(vecambiguous)
FAP = np.array(vecfap)
FBP = np.array(vecfbp)

#Calculate the difference to get the direction of oscilatory sine wave (in the double well figure imagine the x axis as this difference and the y as the energy)
diff_pattern = FAP - FBP
#Note:  diff[i] =  1 -> Unique to FAP
# diff[i] = -1 -> Unique to FBP
# diff[i] =  0 -> Shared by both
gamma = 0.0   # ← controls well asymmetry : We can make the wells equal at gamma = 0
bias_vector = gamma * diff_pattern
# Tuning parameters (Adjusted for Hebbian weight scaling)
H0 = 0.5              #amplitude of oscillating input signal for stochastic resonance
bias_strength = 0.1   #How hard the ambiguous input drives the network? (Looking at the input vectors this pushes both equally)
f_signal=0.0023  #Can be varied but I took this value from "Stochastic Resonance in the Perceptual Interpretation of Ambiguous Figures: A Neural Network Model by Riami et al 1994" which is my comparision

W = np.zeros((N, N))   #WEIGHT MATRIX for Hebbian Learning

# Hopfield networks learn best with bipolar states (-1, 1)
#We map our (0, 1) type patterns to (-1, 1) to calculate the synapses.
bipolar_FAP=2*FAP-1
bipolar_FBP =  2*FBP-1

for i in range(N):
    for j in range(N):
        if i==j:
            continue # No self-connections
        #Hebbian rule: W_ij = (1/N) * Sum(Pattern^mu_i * Pattern^mu_j)
        W[i, j] = (bipolar_FAP[i]*bipolar_FAP[j]+bipolar_FBP[i]*bipolar_FBP[j])/N #Outer product

#SIMULATION using weight MATRIX
def run_simulation(T,f_signal,sweeps=200000):
    period=int(round(1.0/f_signal))
    omega =2.0 *math.pi / period
    #the State Vector tracks all 16 neurons
    S=np.copy(FAP)#start perfectly in FAP's basin

    current_state = 1
    samples = np.zeros(sweeps, dtype=np.int8)

    for sweep in range(sweeps):
        H_t = H0 * math.sin(omega * sweep)

        #Asynchronous update per sweep
        for _ in range(N):
            #pick 1 neuron randomly (0 to 15 in index)
            i = random.randrange(N)

            #The Local Field Calculation, basically will it switch?
            # h_i = Network influence + External Ambiguous Bias (the regular necker's cube as an input) + Oscillating Field
            #We can map S_j to bipolar (2*S - 1) for the symmetric network dynamics
            network_influence = np.dot(W[i],(2 * S - 1))
            external_bias=bias_strength*ambiguous[i]

            #we use H_t to only pushes on segments unique to FAP or FBP this pushing it b/w these 2 only
            oscillating_attention = H_t * diff_pattern[i]
            h_i = network_influence + external_bias + oscillating_attention + bias_vector[i]#bias vector is the one with gamm*diff

            #Calculate Energy Change (dE) to flip this specific neuron
            # if S[i] is 1, then turning it to 0 means a change (dS) of -1.
            # If S[i] is 0, turning it to 1 means a change (dS) of +1.
            dS = 1 - 2 * S[i]
            dE = -h_i * dS    #standard Hopfield Hamiltonian function

            #Metropolis-Hastings (Glauber Dynamics) flip decision -> https://en.wikipedia.org/wiki/Glauber_dynamics
            if T==0:
                if dE<=0:
                    S[i]=1-S[i]
            else:
                if dE<=0 or random.random()<math.exp(-dE/T):
                    S[i] = 1-S[i]

        #This is reminiscent of physics with a Macroscopic State depending on past patterns (Hysteresis)
        #We now simply count the active states based on the number of "unique" segments of FAP or FBP
        unique_fap_active=np.sum(S[diff_pattern==1])   # Max of 5
        unique_fbp_active = np.sum(S[diff_pattern==-1])  #5 are diff

        total_unique = np.sum(diff_pattern != 0)
        m = (unique_fap_active - unique_fbp_active) / total_unique
        threshold = 0.18  # tune this
        if m > threshold:
            current_state = 1      # FAP
        elif m < -threshold:
            current_state = -1     # FBP
        else:
            current_state = 0      # UNDEFINED

        samples[sweep]=current_state

    burn_in=max(5000, sweeps // 10)
    return samples[burn_in:], period


# POWER SPECTRUM - Used chatgpt for this to use the fourier transform properly
def compute_psd(x, period, f_signal):
    x = x - np.mean(x)
    nperseg = period * 128 # Adjusted to prevent array overflow on shorter sweeps

    freqs, power = welch(x, fs=1.0, nperseg=nperseg, scaling='density')

    peak_bins=set()
    max_harmonics=30

    for h in range(1, max_harmonics + 1):
        target_f =h*f_signal
        bin_idx=np.argmin(np.abs(freqs-target_f))
        peak_bins.update([bin_idx-1,bin_idx,bin_idx+1])

    background = np.copy(power)
    for h in range(1,max_harmonics+1):
        target_f=h*f_signal
        bin_idx=np.argmin(np.abs(freqs - target_f))
        left=max(0, bin_idx-2)
        right=min(len(power)-1, bin_idx + 2)
        background[left:right+1]=np.linspace(background[left],background[right],right-left+1)

    w=11
    smoothed_background=np.convolve(background,np.ones(w)/w,mode='same')

    final_power=np.copy(smoothed_background)#for ease of viewing
    for idx in sorted(list(peak_bins)):
        if 0 <= idx<len(final_power):
            final_power[idx]=power[idx]

    return freqs,final_power

#plot the results of all of this! To show resonance also occurs at certain noise levels

def switching_rate(samples):
    switches = np.sum(samples[1:] != samples[:-1])
    return switches / len(samples)
def dwell_times(samples):
    dwells = []
    current = samples[0]
    count = 1

    for s in samples[1:]:
        if s == current:
            count += 1
        else:
            dwells.append(count)
            count = 1
            current = s

    dwells.append(count)
    return np.array(dwells)
def mean_dwell(samples):
    return np.mean(dwell_times(samples))
def percept_bias(samples):
    return {
        1: np.mean(samples == 1),
        0: np.mean(samples == 0),
        -1: np.mean(samples == -1)
    }
def dwell_by_state_full(samples):
    dwells =[]
    states = []

    current = samples[0]
    count = 1

    for s in samples[1:]:
        if s == current:
            count += 1
        else:
            dwells.append(count)
            states.append(current)
            current = s
            count = 1

    dwells.append(count)
    states.append(current)

    return np.array(dwells), np.array(states)
def summarize_dwell_states(samples):
    dwells, states = dwell_by_state_full(samples)

    summary = {}

    for s in [1, 0, -1]:   # enforce all states appear
        s_dwells = dwells[states == s]

        if len(s_dwells) > 0:
            summary[s] = {
                "mean": np.mean(s_dwells),
                "count": len(s_dwells),
                "total_time": np.sum(s_dwells)
            }
        else:
            summary[s] = {
                "mean": np.nan,
                "count": 0,
                "total_time": 0
            }

    return summary
def undefined_fraction(samples):
    return np.mean(samples == 0)
def tempfilter(signal):
    signal=np.array(signal)
    n=len(signal)
    for i in range(3, n-3):
        if (signal[i-3:i]==1).all() and signal[i]==0 and (signal[i+1:i+4]==1).all():
            signal[i] = 1
        elif (signal[i-3:i]==-1).all() and signal[i]==0 and (signal[i+1:i+4] == -1).all():
            signal[i] = -1
        # Check if previous are -1s and next are 1s (and vice versa)
        elif (signal[i-3:i] == -1).all() and signal[i] == 0 and (signal[i+1:i+4] == 1).all():
            signal[i] = 1
        elif (signal[i-3:i] == 1).all() and signal[i] == 0 and (signal[i+1:i+4] == -1).all():
            signal[i] = -1

    return signal
if __name__ == "__main__":
    #temperatures lowered because the Hebbian weights W are smaller in absolute value compared to the Ria97 paper reference before with a manual 0.02 / 0.04 matrix
    temperatures = [0.00,0.15,0.35,0.60]
    spectra = {}

    for T in temperatures:
        print(f"running for T = {T}   ...")
        x, period = run_simulation(T,f_signal, sweeps=500000)
        x=tempfilter(x)
        freqs, power = compute_psd(x,period, f_signal)
        spectra[T] = (freqs, power)
        summary = summarize_dwell_states(x)
        print(f"Switching rate: {switching_rate(x):.5f}")
        print(f"Mean dwell time: {mean_dwell(x):.2f}")

        bias = percept_bias(x)
        print("\nPercept bias:")
        print(f"  +1 (FAP): {bias[1]:.3f}")
        print(f"   0 (Undefined): {bias[0]:.3f}")
        print(f"  -1 (FBP): {bias[-1]:.3f}")

        print(f"\nUndefined fraction: {undefined_fraction(x):.3f}")

        print("\n -- DWELL BY STATE --")
        for s in [1, 0, -1]:
            print(f"State {s}:")
            print(f"  Mean dwell: {summary[s]['mean']}")
            print(f"  Count: {summary[s]['count']}")
            print(f"  Total time: {summary[s]['total_time']}")
            print()
        dw = dwell_times(x)
        plt.hist(dw, bins=50, density=True)
        plt.xlabel("Dwell time")
        plt.ylabel("Probability")
        plt.title("Dwell time distribution")
        plt.show()

    plt.figure(figsize=(8, 5))

    for T, (f, p) in spectra.items():
        mask=f>0
        if abs(T - 0.15) < 0.01:
            style ='--'
        elif abs(T - 0.35) < 0.01:
            style=":"
        elif abs(T - 0.60) < 0.01:
            style='-.'
        else:
            style='-'
        plt.loglog(f[mask],p[mask],style, label=f"T = {T}", linewidth=1.5)

    plt.xlabel("frequency")
    plt.ylabel("P.S.D. (power spectrum)")
    plt.title("Power Spectra with the necker's cube vectorised")
    plt.legend()
    plt.grid(True,which="both", ls="--", alpha=0.5)
    plt.xlim(1e-4,5e0)
    plt.ylim(1e-5,1e5)
    plt.show()
```
